```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the e-commerce dataset

shop_data = pd.read_csv('pakistani_ecommerce_dataset.csv')

# Preview first five rows

shop_data.head()

# Q1: Which product category appears most frequently?

top_category = shop_data['product_category'].mode()[0]
print("Most frequent product category:", top_category)

# Q2: What is the average order value?

avg_order_value = shop_data['order_value'].sum() / shop_data['order_value'].count()
print("Average order value:", avg_order_value)

# Q3: Which city has the highest number of orders?

top_city = shop_data['city'].mode()[0]
print("City with most orders:", top_city)

# Q4: Which payment method is used most often?

top_payment = shop_data['payment_method'].mode()[0]
print("Most used payment method:", top_payment)

# Q5: Visualize average order value across cities

city_avg = shop_data.groupby('city', as_index=False)['order_value'].mean()
city_avg = city_avg.sort_values(by='order_value', ascending=False)

# Draw the bar chart

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(city_avg['city'], city_avg['order_value'], color=sns.color_palette('deep')[0])
ax.set_title('Average Order Value by City')
ax.set_xlabel('City')
ax.set_ylabel('Average Order Value')
ax.set_xticklabels(city_avg['city'], rotation=45, ha='right')
fig.tight_layout()
plt.show()
```

```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the student lifestyle dataset

student_data = pd.read_csv('student_lifestyle_dataset.csv')

# Preview first five rows

student_data.head()

# Q1: What is the mean number of study hours?

mean_study = student_data['study_hours'].sum() / len(student_data['study_hours'])
print("Mean study hours:", mean_study)

# Q2: Average sleep hours grouped by gender

sleep_by_gender = student_data.groupby('gender', as_index=False)['sleep_hours'].mean()
print(sleep_by_gender)

# Q3: Correlation between screen time and sleep hours

correlation_matrix = student_data[['screen_time', 'sleep_hours']].corr()
print(correlation_matrix)

# Q4: Histogram of daily transport cost

fig, ax = plt.subplots(figsize=(10, 6))
sns.histplot(
    data=student_data,
    x='daily_transport_cost',
    bins=15,
    kde=True,
    ax=ax
)
ax.set_title('Distribution of Daily Transport Cost')
ax.set_xlabel('Daily Transport Cost')
ax.set_ylabel('Frequency')
fig.tight_layout()
plt.show()
```
